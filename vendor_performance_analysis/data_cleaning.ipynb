```python
vendor_summary = (
    purchases.groupby(["VendorNumber", "VendorName"])
    .agg(
        TotalPurchaseDollars=("Dollars", "sum"),
        TotalPurchaseQuantity=("Quantity", "sum"),
        PurchaseOrders=("PONumber", "nunique"),
        AvgPurchasePrice=("PurchasePrice", "mean")
    )
    .reset_index()
)

vendor_summary.head()
```

```python
vendor_sales = (
    sales.groupby("VendorNo")
    .agg(
        TotalSalesDollars=("SalesDollars", "sum"),
        TotalSalesQuantity=("SalesQuantity", "sum"),
        AvgSalesPrice=("SalesPrice", "mean")
    )
    .reset_index()
    .rename(columns={"VendorNo": "VendorNumber"})
)

vendor_summary = vendor_summary.merge(
    vendor_sales,
    on="VendorNumber",
    how="left"
)

vendor_summary.head()
```

```python
end_inventory["InventoryValue"] = (
    end_inventory["onHand"] * end_inventory["Price"]
)

ending_inventory = (
    end_inventory.groupby("Store")
    .agg(
        EndingInventoryUnits=("onHand", "sum"),
        EndingInventoryValue=("InventoryValue", "sum")
    )
    .reset_index()
)

store_sales = (
    sales.groupby("Store")
    .agg(
        SalesDollars=("SalesDollars", "sum"),
        SalesQuantity=("SalesQuantity", "sum")
    )
    .reset_index()
)

store_performance = ending_inventory.merge(
    store_sales,
    on="Store",
    how="left"
)

store_performance["InventoryToSalesRatio"] = (
    store_performance["EndingInventoryValue"]
    / store_performance["SalesDollars"]
)

store_performance.sort_values(
    "InventoryToSalesRatio",
    ascending=False
).head(10)
```

```python
freight_summary = (
    vendor_invoice.groupby("VendorNumber")
    .agg(
        TotalFreight=("Freight", "sum"),
        TotalInvoiceDollars=("Dollars", "sum"),
        InvoiceCount=("PONumber", "nunique")
    )
    .reset_index()
)

freight_summary["FreightPct"] = (
    freight_summary["TotalFreight"]
    / freight_summary["TotalInvoiceDollars"] * 100
)

freight_summary.head()
```

```python
vendor_summary = vendor_summary.merge(
    freight_summary,
    on="VendorNumber",
    how="left"
)

vendor_summary["TotalFreight"] = (
    vendor_summary["TotalFreight"].fillna(0)
)

vendor_summary["FreightPct"] = (
    vendor_summary["FreightPct"].fillna(0)
)
```

```python
vendor_summary["AdjustedProfit"] = (
    vendor_summary["GrossProfit"]
    - vendor_summary["TotalFreight"]
)

vendor_summary["AdjustedMarginPct"] = (
    vendor_summary["AdjustedProfit"]
    / vendor_summary["TotalSalesDollars"] * 100
)
```

```python
vendor_performance = vendor_summary[
    [
        "VendorNumber",
        "VendorName",
        "TotalPurchaseDollars",
        "TotalPurchaseQuantity",
        "PurchaseOrders",
        "TotalSalesDollars",
        "TotalSalesQuantity",
        "GrossProfit",
        "ProfitMarginPct",
        "TotalFreight",
        "FreightPct",
        "AdjustedProfit",
        "AdjustedMarginPct",
        "SalesToPurchaseRatio"
    ]
].copy()

vendor_performance = vendor_performance.sort_values(
    "AdjustedProfit",
    ascending=False
)

vendor_performance.head(10)
```

```python
# REBUILD vendor_summary from scratch using VendorNumber as the key

# 1. Purchases
purchase_summary = (
    purchases.groupby("VendorNumber")
    .agg(
        TotalPurchaseDollars=("Dollars", "sum"),
        TotalPurchaseQuantity=("Quantity", "sum"),
        PurchaseOrders=("PONumber", "nunique"),
        AvgPurchasePrice=("PurchasePrice", "mean")
    )
    .reset_index()
)

# 2. One display name per vendor number
vendor_names = (
    purchases.groupby("VendorNumber")["VendorName"]
    .first()
    .reset_index()
)

purchase_summary = purchase_summary.merge(
    vendor_names,
    on="VendorNumber",
    how="left"
)

# 3. Sales
vendor_sales = (
    sales.groupby("VendorNo")
    .agg(
        TotalSalesDollars=("SalesDollars", "sum"),
        TotalSalesQuantity=("SalesQuantity", "sum"),
        AvgSalesPrice=("SalesPrice", "mean")
    )
    .reset_index()
    .rename(columns={"VendorNo": "VendorNumber"})
)

# 4. Merge purchases + sales
vendor_summary = purchase_summary.merge(
    vendor_sales,
    on="VendorNumber",
    how="left"
)

# 5. Calculate KPIs
vendor_summary["GrossProfit"] = (
    vendor_summary["TotalSalesDollars"]
    - vendor_summary["TotalPurchaseDollars"]
)

vendor_summary["ProfitMarginPct"] = (
    vendor_summary["GrossProfit"]
    / vendor_summary["TotalSalesDollars"] * 100
)

vendor_summary["SalesToPurchaseRatio"] = (
    vendor_summary["TotalSalesDollars"]
    / vendor_summary["TotalPurchaseDollars"]
)

# Check Vendor 2000
vendor_summary[
    vendor_summary["VendorNumber"] == 2000
]
```

```python
vendor_summary = vendor_summary.merge(
    freight_summary,
    on="VendorNumber",
    how="left"
)

vendor_summary["TotalFreight"] = vendor_summary["TotalFreight"].fillna(0)
vendor_summary["FreightPct"] = vendor_summary["FreightPct"].fillna(0)

vendor_summary["AdjustedProfit"] = (
    vendor_summary["GrossProfit"] - vendor_summary["TotalFreight"]
)

vendor_summary["AdjustedMarginPct"] = (
    vendor_summary["AdjustedProfit"]
    / vendor_summary["TotalSalesDollars"] * 100
)
```

```python
vendor_performance = vendor_summary[
    [
        "VendorNumber",
        "VendorName",
        "TotalPurchaseDollars",
        "TotalPurchaseQuantity",
        "PurchaseOrders",
        "TotalSalesDollars",
        "TotalSalesQuantity",
        "GrossProfit",
        "ProfitMarginPct",
        "TotalFreight",
        "FreightPct",
        "AdjustedProfit",
        "AdjustedMarginPct",
        "SalesToPurchaseRatio"
    ]
].copy()

vendor_performance = vendor_performance.sort_values(
    "AdjustedProfit",
    ascending=False
)

vendor_performance.head(10)
```

```python
query = """
WITH purchase_summary AS (
    SELECT
        vendornumber,
        TRIM(vendorname) AS vendorname,
        SUM(dollars) AS total_purchase_dollars,
        SUM(quantity) AS total_purchase_quantity,
        COUNT(DISTINCT ponumber) AS purchase_orders
    FROM purchases
    GROUP BY vendornumber, TRIM(vendorname)
),

sales_summary AS (
    SELECT
        vendorno AS vendornumber,
        SUM(salesdollars) AS total_sales_dollars,
        SUM(salesquantity) AS total_sales_quantity
    FROM sales
    GROUP BY vendorno
),

freight_summary AS (
    SELECT
        vendornumber,
        SUM(freight) AS total_freight
    FROM vendor_invoice
    GROUP BY vendornumber
)

SELECT
    p.vendornumber,
    p.vendorname,

    ROUND(p.total_purchase_dollars::numeric, 2)
        AS total_purchase_dollars,

    ROUND(s.total_sales_dollars::numeric, 2)
        AS total_sales_dollars,

    ROUND(
        (s.total_sales_dollars - p.total_purchase_dollars)::numeric,
        2
    ) AS gross_profit,

    ROUND(COALESCE(f.total_freight, 0)::numeric, 2)
        AS total_freight,

    ROUND(
        (
            COALESCE(f.total_freight, 0)
            / NULLIF(p.total_purchase_dollars, 0)
            * 100
        )::numeric,
        2
    ) AS freight_pct,

    ROUND(
        (
            s.total_sales_dollars
            - p.total_purchase_dollars
            - COALESCE(f.total_freight, 0)
        )::numeric,
        2
    ) AS adjusted_profit,

    ROUND(
        (
            (
                s.total_sales_dollars
                - p.total_purchase_dollars
                - COALESCE(f.total_freight, 0)
            )
            / NULLIF(s.total_sales_dollars, 0)
            * 100
        )::numeric,
        2
    ) AS adjusted_margin_pct

FROM purchase_summary p

JOIN sales_summary s
    ON p.vendornumber = s.vendornumber

LEFT JOIN freight_summary f
    ON p.vendornumber = f.vendornumber

WHERE s.total_sales_dollars > 0

ORDER BY adjusted_profit DESC
LIMIT 10;
"""

vendor_freight_analysis = pd.read_sql(query, engine)

vendor_freight_analysis
```

```python
final_vendor_analysis = pd.read_sql("""
WITH purchase_summary AS (
    SELECT
        vendornumber,
        TRIM(vendorname) AS vendorname,
        SUM(dollars) AS total_purchase_dollars,
        SUM(quantity) AS total_purchase_quantity,
        COUNT(DISTINCT ponumber) AS purchase_orders,
        AVG(purchaseprice) AS avg_purchase_price
    FROM purchases
    GROUP BY vendornumber, TRIM(vendorname)
),

sales_summary AS (
    SELECT
        vendorno AS vendornumber,
        SUM(salesdollars) AS total_sales_dollars,
        SUM(salesquantity) AS total_sales_quantity,
        AVG(salesprice) AS avg_sales_price
    FROM sales
    GROUP BY vendorno
),

freight_summary AS (
    SELECT
        vendornumber,
        SUM(freight) AS total_freight
    FROM vendor_invoice
    GROUP BY vendornumber
)

SELECT
    p.vendornumber,
    p.vendorname,

    ROUND(p.total_purchase_dollars::numeric, 2)
        AS total_purchase_dollars,

    p.total_purchase_quantity,
    p.purchase_orders,

    ROUND(p.avg_purchase_price::numeric, 2)
        AS avg_purchase_price,

    ROUND(s.total_sales_dollars::numeric, 2)
        AS total_sales_dollars,

    s.total_sales_quantity,

    ROUND(s.avg_sales_price::numeric, 2)
        AS avg_sales_price,

    ROUND(
        (s.total_sales_dollars - p.total_purchase_dollars)::numeric,
        2
    ) AS gross_profit,

    ROUND(
        (
            (s.total_sales_dollars - p.total_purchase_dollars)
            / NULLIF(s.total_sales_dollars, 0) * 100
        )::numeric,
        2
    ) AS profit_margin_pct,

    ROUND(COALESCE(f.total_freight, 0)::numeric, 2)
        AS total_freight,

    ROUND(
        (
            s.total_sales_dollars
            - p.total_purchase_dollars
            - COALESCE(f.total_freight, 0)
        )::numeric,
        2
    ) AS adjusted_profit,

    ROUND(
        (
            (
                s.total_sales_dollars
                - p.total_purchase_dollars
                - COALESCE(f.total_freight, 0)
            )
            / NULLIF(s.total_sales_dollars, 0) * 100
        )::numeric,
        2
    ) AS adjusted_margin_pct

FROM purchase_summary p

JOIN sales_summary s
    ON p.vendornumber = s.vendornumber

LEFT JOIN freight_summary f
    ON p.vendornumber = f.vendornumber

WHERE s.total_sales_dollars > 0

ORDER BY total_sales_dollars DESC;
""", engine)

final_vendor_analysis.head(10)
```
